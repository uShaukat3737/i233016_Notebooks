# Task 3

```python
import pandas as pd
import seaborn as sns

df_titanic = sns.load_dataset("titanic")

print("Missing values before preprocessing:")
print(df_titanic.isnull().sum())

#processing
df_titanic["age"] = df_titanic["age"].fillna(df_titanic["age"].mean())
df_titanic["embarked"] = df_titanic["embarked"].fillna(df_titanic["embarked"].mode()[0])
df_titanic = df_titanic.drop(columns=["deck"])

print("Missing values now:")
print(df_titanic.isnull().sum())
```

# Task 4

```python
# load data
df_iris = sns.load_dataset("iris")
df_titanic = sns.load_dataset("titanic")
df_tips = sns.load_dataset("tips")

#iris
print("Iris before:", df_iris.shape)
print("Duplicate ro", df_iris.duplicated().sum())

df_iris = df_iris.drop_duplicates()
print("Iris after:", df_iris.shape)

#titanic
print("Titanic Before:", df_titanic.shape)
print("Duplicate", df_titanic.duplicated().sum())

df_titanic = df_titanic.drop_duplicates()
print("Titanic after", df_titanic.shape)

#tips
print("Tips Before", df_tips.shape)
print("Duplicate", df_tips.duplicated().sum())

df_tips = df_tips.drop_duplicates()
print("Tips after:", df_tips.shape)
```

# Task 5

```python
#group by species and then apply mean and standard deviation through pandas on the grouped data
iris_stats = df_iris.groupby("species").agg(["mean", "std"], numeric_only=True)
print(iris_stats)

#select total_bill and tips and then apply mean and median through pandas on the grouped data
tips_stats = df_tips[["total_bill", "tip"]].agg(["mean", "median"], numeric_only=True)
print(tips_stats)

# group by survived and then get mean vals for age and fare 
titanic_stats = df_titanic.groupby("survived")[["age", "fare"]].mean()
print(titanic_stats)
```

# Task 7

```python
df_tips = sns.load_dataset("tips")

#25%, 75%, IQR (75-25)
Q1_tb = df_tips["total_bill"].quantile(0.25)
Q3_tb = df_tips["total_bill"].quantile(0.75)
IQR_tb = Q3_tb - Q1_tb

#
lower_tb = Q1_tb - 1.5 * IQR_tb
upper_tb = Q3_tb + 1.5 * IQR_tb
outliers_total_bill = df_tips[(df_tips["total_bill"] < lower_tb) |(df_tips["total_bill"] > upper_tb)]

Q1_tip = df_tips["tip"].quantile(0.25)
Q3_tip = df_tips["tip"].quantile(0.75)
IQR_tip = Q3_tip - Q1_tip

lower_tip = Q1_tip - 1.5 * IQR_tip
upper_tip = Q3_tip + 1.5 * IQR_tip
outliers_tip = df_tips[ (df_tips["tip"] < lower_tip) | (df_tips["tip"] > upper_tip)]

print("Total_bill ", outliers_total_bill.shape[0])
print("Tip outlier", outliers_tip.shape[0])

plt.boxplot(df_tips["total_bill"])
plt.show()
plt.boxplot(df_tips["tip"])
plt.show()
```
