#Task 2

```python
Paths = {
    "W": {"distance": 300, "time": 25, "congestion": "Medium"},
    "X": {"distance": 200, "time": 15, "congestion": "Low"},
    "Y": {"distance": 150, "time": 20, "congestion": "Medium"},
    "Z": {"distance": 250, "time": 10, "congestion": "High"}
}

Congestions={
    'Low':3,
    'Medium':2,
    'High':1,
}

def utilityFunction(path):
    
    distance = path['distance']
    congestion = Congestions[path['congestion']]
    time = path['time']

    return (time*4+distance*3+congestion*1.1)

utilities={}

for path_name, path_info in Paths.items():
    utilities[path_name] = utilityFunction(path_info)

best_path = min(utilities, key=utilities.get)

for path, score in utilities.items():
    print("Path", path, ":", score)

print("\nSelected Path:", best_path, "with utility", utilities[best_path])
```

#Task 3

```python
#environment state 
environment = {
    "Section 1": "Healthy",
    "Section 2": "Wilted"
}

#agent state (unknown at start)
internal_state = {
    "Section 1": "Unknown",
    "Section 2": "Unknown"
}

#agent position
agent_location = "Section 1"

#Agent observes the current section
def observe(section):
    return environment[section]

#Watering plants
def water(section):
    environment[section] = "Healthy"

#change section
def move(section):
    if section == "Section 1":
        return "Section 2"
    
    if section == "Section 2":
        return "Section 1"

# numbre of runs
time_steps = 6

for t in range(time_steps):

    #current iteration 
    current = t+1
    print("Time Step: " , current ) 
    print("Agent location: " + agent_location)

    #initiate
    perception = observe(agent_location)
    internal_state[agent_location] = perception

    #think for action
    if perception == "Wilted":
        action = "Water"
        water(agent_location)
    else:
        action = "Move"
        agent_location = move(agent_location)

    print("Action taken: " + action)
    print("Agent internal state:", internal_state)
    print("Actual environment:", environment)
```
